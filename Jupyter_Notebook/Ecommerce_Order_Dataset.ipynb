```python
# Quantos dados NAN existem
df_Orders.isnull().sum()
```

```python
# mostrando dados nulos
df_Orders[df_Orders["order_estimated_delivery_date"].isnull()]
```

```python
# Verificação final
print(df_Orders.info())
print("\nContagem final de nulos:")
print(df_Orders.isnull().sum())
```
